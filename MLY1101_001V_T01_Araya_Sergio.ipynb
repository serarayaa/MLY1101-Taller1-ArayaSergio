````python
# Librerías base
import requests
import json
import pandas as pd
from pathlib import Path

OUTPUT_DIR = Path('datasets')
OUTPUT_DIR.mkdir(exist_ok=True)

print(f'Carpeta de salida: {OUTPUT_DIR.resolve()}')
````

# Fuente 1 — API 1

**Nombre de la API:** TVMaze API  

**Documentación:** https://www.tvmaze.com/api  

**Endpoint utilizado:** https://api.tvmaze.com/shows

**Descripción de los datos:** Catálogo de producciones televisivas y series de entretenimiento digital global. Incluye títulos, géneros, puntuaciones, estado de emisión y fechas de estreno.
  
**Relación con el objetivo:** Aporta la dimensión de producciones audiovisuales (series y programas de TV) al análisis general de contenidos de entretenimiento y cultura pop.

````python
# CONFIGURACIÓN API 1
API1_URL = 'https://api.tvmaze.com/shows'
API1_MIN_REGISTROS = 200
api1_headers = {}
api1_params = {}
````

````python
# CONSUMO API 1
api1_registros = []

response = requests.get(API1_URL, headers=api1_headers, params=api1_params, timeout=30)
print('Status API 1:', response.status_code)
response.raise_for_status()

data = response.json()

# TODO: extraiga desde 'data' la lista de registros correspondiente.
api1_registros = data  # TVMaze responde directamente con una lista de 250 series

print('Registros API 1:', len(api1_registros))
````

# Fuente 2 — API 2

**Nombre de la API:** SuperHero API (Akabab Mirror)

**Documentación:** https://akabab.github.io/superhero-api/api/

**Endpoint utilizado:** https://akabab.github.io/superhero-api/api/all.json

**Descripción de los datos:** Contiene información detallada sobre cientos de superhéroes y villanos de múltiples universos y franquicias (Marvel, DC, etc.).

**Relación con el objetivo:** Permite perfilar y categorizar los personajes centrales de las producciones cinematográficas y franquicias del entretenimiento abordadas en la API 1, ampliando el análisis hacia sus atributos individuales.

````python
# CONFIGURACIÓN API 2
API2_URL = 'https://akabab.github.io/superhero-api/api/all.json'
API2_MIN_REGISTROS = 200
api2_headers = {}
api2_params = {}
````

````python
# CONSUMO API 2
api2_registros = []

response = requests.get(API2_URL, headers=api2_headers, params=api2_params, timeout=30)
print('Status API 2:', response.status_code)
response.raise_for_status()

data = response.json()

# TODO: extraiga la lista de registros.
api2_registros = data

print('Registros API 2:', len(api2_registros))
````

````python
# GUARDAR DATASET API 2
API2_ARCHIVO = OUTPUT_DIR / 'dataset_api_2.json'

with open(API2_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api2_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API2_ARCHIVO)
print('Registros guardados:', len(api2_registros))
````

# Fuente 3 — API 3

**Nombre de la API:** PokeAPI

**Documentación:** https://pokeapi.co/docs/v2  

**Endpoint utilizado:** https://pokeapi.co/api/v2/pokemon  

**Descripción de los datos:** Catálogo de personajes y entidades pertenecientes a la franquicia global Pokémon.

**Relación con el objetivo:** Representa al sector del entretenimiento interactivo y franquicias multimedia dentro del análisis general de cultura pop.

````python
# CONFIGURACIÓN API 3
API3_URL = 'https://pokeapi.co/api/v2/pokemon'
API3_MIN_REGISTROS = 200
api3_headers = {}
api3_params = {
    'limit': 250
}
````

````python
# CONSUMO API 3
api3_registros = []

response = requests.get(API3_URL, headers=api3_headers, params=api3_params, timeout=30)
print('Status API 3:', response.status_code)
response.raise_for_status()

data = response.json()

# TODO: extraiga la lista de registros.
api3_registros = data['results']

print('Registros API 3:', len(api3_registros))
````

````python
# GUARDAR DATASET API 3
API3_ARCHIVO = OUTPUT_DIR / 'dataset_api_3.json'

with open(API3_ARCHIVO, 'w', encoding='utf-8') as f:
    json.dump(api3_registros, f, ensure_ascii=False, indent=2)

print('Archivo generado:', API3_ARCHIVO)
print('Registros guardados:', len(api3_registros))
````
